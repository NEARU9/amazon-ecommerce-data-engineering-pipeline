## 4. Data Cleaning

```python
df_clean = df.copy()

print("Size data:", df_clean.shape)
```

```python
df_clean['purchase_date'] = pd.to_datetime(df_clean['purchase_date'])

print(df_clean['purchase_date'].dtype)
```

```python
categorical_columns = [
    'category',
    'subcategory',
    'brand',
    'location',
    'device',
    'payment_method',
    'delivery_status'
]

for col in categorical_columns:
    df_clean[col] = df_clean[col].str.strip()
```

## 5. Data Transformation

```python
# Membuat kolom turunan dari purchase_date

df_clean['purchase_year'] = df_clean['purchase_date'].dt.year
df_clean['purchase_month'] = df_clean['purchase_date'].dt.month
df_clean['purchase_month_name'] = df_clean['purchase_date'].dt.month_name()
df_clean['purchase_quarter'] = df_clean['purchase_date'].dt.quarter
df_clean['purchase_day'] = df_clean['purchase_date'].dt.day
df_clean['purchase_day_name'] = df_clean['purchase_date'].dt.day_name()
```

```python
df_clean[
    [
        'purchase_date',
        'purchase_year',
        'purchase_month',
        'purchase_month_name',
        'purchase_quarter',
        'purchase_day',
        'purchase_day_name'
    ]
].head()
```

```python
df_clean['revenue'] = df_clean['final_price']
```

```python
df_clean[['price', 'discount', 'final_price', 'revenue']].head()
```

```python
df_clean['discount_amount'] = (
    df_clean['price'] - df_clean['final_price']
)
```

```python
df_clean[
    ['price', 'discount', 'discount_amount', 'final_price']
].head()
```

```python
def classify_shipping(days):
    if days <= 2:
        return 'Fast'
    elif days <= 4:
        return 'Standard'
    else:
        return 'Slow'

df_clean['shipping_category'] = (
    df_clean['shipping_time_days']
    .apply(classify_shipping)
)
```

```python
df_clean['shipping_category'].value_counts()
```

```python
def classify_rating(rating):
    if rating < 3:
        return 'Low'
    elif rating < 4:
        return 'Medium'
    else:
        return 'High'

df_clean['rating_category'] = (
    df_clean['rating']
    .apply(classify_rating)
)
```

```python
df_clean['rating_category'].value_counts()
```

```python
def classify_seller_rating(rating):
    if rating < 3:
        return 'Low'
    elif rating < 4:
        return 'Medium'
    else:
        return 'High'

df_clean['seller_rating_category'] = (
    df_clean['seller_rating']
    .apply(classify_seller_rating)
)
```

```python
df_clean['return_flag'] = (
    df_clean['is_returned']
    .astype(int)
)
```

```python
return_rate = df_clean['return_flag'].mean()

print("Return rate:", return_rate)
```

```python
df_clean['delivered_flag'] = (
    df_clean['delivery_status'] == 'Delivered'
).astype(int)
```

```python
df_clean['delayed_flag'] = (
    df_clean['delivery_status'] == 'Delayed'
).astype(int)
```

## 6. Data Modeling

```python
# ==========================================
# DIM USER
# ==========================================

dim_user = (
    df_clean[
        [
            'user_id',
            'location',
            'device'
        ]
    ]
    .drop_duplicates('user_id')
    .reset_index(drop=True)
)

# Surrogate key
dim_user.insert(0, 'user_key', range(1, len(dim_user) + 1))

print(dim_user.shape)
dim_user.head()
```

```python
# ==========================================
# DIM PRODUCT
# ==========================================

dim_product = (
    df[
        [
            'product_id',
            'category',
            'subcategory',
            'brand'
        ]
    ]
    .drop_duplicates('product_id')
    .reset_index(drop=True)
)

dim_product.insert(
    0,
    'product_key',
    range(1, len(dim_product) + 1)
)

print(dim_product.shape)
dim_product.head()
```

```python
# ==========================================
# DIM SELLER
# ==========================================

dim_seller = (
    df_clean[
        [
            'seller_id',
            'seller_rating',
            'seller_rating_category'
        ]
    ]
    .drop_duplicates('seller_id')
    .reset_index(drop=True)
)

dim_seller.insert(
    0,
    'seller_key',
    range(1, len(dim_seller) + 1)
)

print(dim_seller.shape)
dim_seller.head()
```

```python
# ==========================================
# DIM DATE
# ==========================================

dim_date = (
    df_clean[
        [
            'purchase_date',
            'purchase_year',
            'purchase_month',
            'purchase_month_name',
            'purchase_quarter',
            'purchase_day',
            'purchase_day_name'
        ]
    ]
    .drop_duplicates('purchase_date')
    .sort_values('purchase_date')
    .reset_index(drop=True)
)

dim_date.insert(
    0,
    'date_key',
    range(1, len(dim_date) + 1)
)

print(dim_date.shape)
dim_date.head()
```

```python
fact_sales = df_clean[
    [
        'user_id',
        'product_id',
        'seller_id',
        'purchase_date',
        'price',
        'discount',
        'discount_amount',
        'final_price',
        'revenue',
        'rating',
        'review_count',
        'stock',
        'shipping_time_days',
        'shipping_category',
        'payment_method',
        'device',
        'delivery_status',
        'is_returned',
        'return_flag',
        'delivered_flag',
        'delayed_flag'
    ]
].copy()
```

```python
#Ganti ID dengan Surrogate Key

#User
fact_sales = fact_sales.merge(
    dim_user[['user_key', 'user_id']],
    on='user_id',
    how='left'
)

#Product
fact_sales = fact_sales.merge(
    dim_product[['product_key', 'product_id']],
    on='product_id',
    how='left'
)

#Seller
fact_sales = fact_sales.merge(
    dim_seller[['seller_key', 'seller_id']],
    on='seller_id',
    how='left'
)

#Date
fact_sales = fact_sales.merge(
    dim_date[['date_key', 'purchase_date']],
    on='purchase_date',
    how='left'
)
```

```python
fact_sales = fact_sales[
    [
        'user_key',
        'product_key',
        'seller_key',
        'date_key',
        'price',
        'discount',
        'discount_amount',
        'final_price',
        'revenue',
        'rating',
        'review_count',
        'stock',
        'shipping_time_days',
        'shipping_category',
        'payment_method',
        'device',
        'delivery_status',
        'is_returned',
        'return_flag',
        'delivered_flag',
        'delayed_flag'
    ]
]
```

```python
fact_sales.insert(
    0,
    'sales_key',
    range(1, len(fact_sales) + 1)
)
```

```python
print("dim_user:", dim_user.shape)
print("dim_product:", dim_product.shape)
print("dim_seller:", dim_seller.shape)
print("dim_date:", dim_date.shape)
print("fact_sales:", fact_sales.shape)
```

## 7. Save Data

```python
print("\n" + "=" * 60)
print("7. SAVE DATA TO CSV")
print("=" * 60)

# Create directories
os.makedirs('../data/staging', exist_ok=True)
os.makedirs('../data/warehouse', exist_ok=True)

# 7.1 Save staging data (cleaned & transformed data)
staging_path = '../data/staging/ecommerce_staging.csv'
df_clean.to_csv(staging_path, index=False)
print(f"✓ Staging data saved to: {staging_path}")
print(f"  Shape: {df_clean.shape}")

# 7.2 Save dimension tables
dim_tables = {
    'dim_user': dim_user,
    'dim_product': dim_product,
    'dim_seller': dim_seller,
    'dim_date': dim_date
}

for name, table in dim_tables.items():
    filepath = f'../data/warehouse/{name}.csv'
    table.to_csv(filepath, index=False)
    print(f"✓ {name} saved to: {filepath}")
    print(f"  Shape: {table.shape}")

# 7.3 Save fact table
fact_path = '../data/warehouse/fact_sales.csv'
fact_sales.to_csv(fact_path, index=False)
print(f"✓ fact_sales saved to: {fact_path}")
print(f"  Shape: {fact_sales.shape}")
```
